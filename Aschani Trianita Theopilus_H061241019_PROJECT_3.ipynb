## 2. Teori singkat dan sumber data

Observatorium geomagnet mengukur medan magnet bumi terus-menerus dengan magnetometer. Hasilnya dinyatakan dalam empat besaran, semuanya dalam nanotesla (nT).

- X adalah komponen ke arah utara geografis.
- Y adalah komponen ke arah timur geografis.
- Z adalah komponen vertikal, positif ke arah bawah.
- F adalah intensitas total medan.

Komponen horizontal H dihitung sebagai akar dari jumlah kuadrat X dan Y. Komponen inilah yang paling jelas memperlihatkan badai di lintang rendah.

Saat badai geomagnet, gumpalan plasma dari Matahari menekan dan mengganggu magnetosfer bumi. Pada awal badai, medan di permukaan biasanya naik tiba-tiba, yang disebut sudden commencement. Setelah itu arus cincin di sekitar bumi menguat sehingga komponen horizontal di lintang rendah dan menengah turun, kadang sampai ratusan nanotesla. Besar penurunan ini diringkas dalam indeks Dst. Karena Guam berada di lintang rendah, rekamannya cocok untuk melihat pengaruh arus cincin.

Konvensi yang dipakai: waktu dalam UTC, medan dalam nT, X ke utara, Y ke timur, dan Z ke bawah.

Data diambil dari layanan web USGS Geomagnetism Program yang terbuka tanpa kata sandi. Jenis data yang tersedia adalah definitive (final), quasi-definitive, adjusted, dan variation (data awal). Recipe mencoba jenis bermutu tertinggi lebih dulu dan mencatat jenis mana yang akhirnya dipakai.

Sumber:
- Dokumentasi layanan data: https://www.usgs.gov/tools/web-service-geomagnetism-data

## 4. Persiapan dan parameter

Semua pilihan yang bisa diubah dikumpulkan di satu tempat. Keluaran dipisah ke tiga folder: mentah untuk respons asli dari layanan, hasil untuk tabel dan laporan, serta gambar untuk grafik.

Garis dasar dipilih pada malam hari waktu setempat Guam (UTC+10), yaitu sekitar pukul 22.00 sampai 03.00, karena variasi harian akibat sinar Matahari paling kecil pada jam itu. Pilihan ini sederhana dan dapat digeser bila rekaman menunjukkan gangguan pada jendela tersebut.

```python
import hashlib
import json
import platform
import subprocess
import sys
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import requests

# Stasiun dan periode
KODE_STASIUN = "GUA"
NAMA_STASIUN = "Guam"
WAKTU_MULAI = pd.Timestamp("2024-05-09 00:00", tz="UTC")
WAKTU_SELESAI = pd.Timestamp("2024-05-13 23:59", tz="UTC")
UNSUR = ["X", "Y", "Z", "F"]
INTERVAL_DETIK = 60

# Jendela garis dasar (malam hari di Guam, sebelum badai tiba)
DASAR_MULAI = pd.Timestamp("2024-05-09 12:00", tz="UTC")
DASAR_SELESAI = pd.Timestamp("2024-05-09 17:00", tz="UTC")

# Hari yang dicari awal badainya
CARI_MULAI = pd.Timestamp("2024-05-10 00:00", tz="UTC")
CARI_SELESAI = pd.Timestamp("2024-05-10 23:59", tz="UTC")

# Batas pemeriksaan mutu (untuk memberi tanda, bukan untuk menghapus)
BATAS_F = (20000, 70000)       # nT
BATAS_H = (5000, 60000)        # nT
AMBANG_LONJAKAN = 30           # nT per menit, ditinjau bila terlampaui

URL_DATA = "https://geomag.usgs.gov/ws/data/"
PERCOBAAN_MAKS = 3

# Folder keluaran
OUT_DIR = Path("hasil_recipe_geomagnet")
MENTAH = OUT_DIR / "mentah"
HASIL = OUT_DIR / "hasil"
GAMBAR = OUT_DIR / "gambar"
for folder in (MENTAH, HASIL, GAMBAR):
    folder.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.3})


def teks_waktu(waktu):
    # Format waktu yang diminta oleh layanan USGS
    return waktu.strftime("%Y-%m-%dT%H:%M:%SZ")


def sha256_berkas(jalur):
    # Sidik jari berkas untuk memastikan berkas tidak berubah
    return hashlib.sha256(Path(jalur).read_bytes()).hexdigest()


def simpan_gambar(fig, nama):
    fig.savefig(GAMBAR / nama, dpi=200, bbox_inches="tight")
```

## 5. Unduh data aktual dari USGS (R-RET-001)

Pengambilan data memakai batas waktu tunggu dan percobaan ulang yang terbatas. Permintaan yang ditolak layanan (kode 4xx) tidak diulang, sedangkan gangguan sementara (kode 5xx, batas laju 429, atau sambungan terputus) diulang paling banyak tiga kali. Respons mentah disimpan apa adanya, lalu catatan pengambilan ditulis terpisah: waktu unduh, alamat akhir, parameter, ukuran, dan sidik jari berkas.

Bila jenis data bermutu tertinggi belum tersedia atau kosong, jenis berikutnya dicoba. Pilihan jenis ini dinyatakan sejak awal dan dicatat pada riwayat pengambilan. Jika semuanya gagal, jangan mengganti data dengan angka buatan; periksa pesan kesalahan dan sambungan internet.

```python
class PermintaanDitolak(Exception):
    pass


def ambil(parameter):
    # GET dengan batas waktu tunggu dan percobaan ulang terbatas
    for ke in range(1, PERCOBAAN_MAKS + 1):
        try:
            jawaban = requests.get(
                URL_DATA,
                params=parameter,
                timeout=120,
                headers={"User-Agent": "Python-Geophysics-Student/1.0"},
            )
        except (requests.Timeout, requests.ConnectionError) as galat:
            print(f"  percobaan {ke}: gangguan sambungan ({galat.__class__.__name__})")
            time.sleep(3 * ke)
            continue

        kode_http = jawaban.status_code
        if kode_http == 429 or kode_http >= 500:
            tunggu = str(jawaban.headers.get("Retry-After", ""))
            tunggu = min(int(tunggu), 60) if tunggu.isdigit() else 3 * ke
            print(f"  percobaan {ke}: layanan menjawab {kode_http}, menunggu {tunggu} detik")
            time.sleep(tunggu)
            continue
        if kode_http >= 400:
            raise PermintaanDitolak(f"status {kode_http}, permintaan tidak diulang")
        return jawaban
    raise RuntimeError("Layanan tidak menjawab setelah percobaan terbatas")


def ubah_ke_tabel(isi):
    # Mengubah jawaban JSON menjadi tabel dengan indeks waktu UTC
    waktu = pd.to_datetime(isi["times"], utc=True)
    tabel = pd.DataFrame(index=waktu)
    for unsur in isi["values"]:
        nama = unsur.get("id") or unsur["metadata"]["element"]
        nilai = pd.to_numeric(pd.Series(unsur["values"]), errors="coerce")
        tabel[nama] = nilai.to_numpy()
    return tabel


jenis_data = ["definitive", "quasi-definitive", "adjusted", "variation"]

data_mentah = None
jawaban_dipakai = None
isi_dipakai = {}
parameter_dipakai = {}
jenis_dipakai = None
riwayat_jenis = []

for jenis in jenis_data:
    parameter = {
        "id": KODE_STASIUN,
        "starttime": teks_waktu(WAKTU_MULAI),
        "endtime": teks_waktu(WAKTU_SELESAI),
        "elements": ",".join(UNSUR),
        "sampling_period": INTERVAL_DETIK,
        "type": jenis,
        "format": "json",
    }
    try:
        jawaban = ambil(parameter)
        isi = jawaban.json()
        tabel = ubah_ke_tabel(isi)
    except Exception as galat:
        print(f"Jenis data {jenis} tidak dipakai: {galat}")
        riwayat_jenis.append({"jenis": jenis, "hasil": f"gagal: {galat}"})
        continue

    bagian_terisi = tabel["X"].notna().mean() if "X" in tabel else 0
    if bagian_terisi >= 0.5:
        riwayat_jenis.append({"jenis": jenis, "hasil": "dipakai"})
        data_mentah, jawaban_dipakai, isi_dipakai = tabel, jawaban, isi
        parameter_dipakai, jenis_dipakai = parameter, jenis
        break
    print(f"Jenis data {jenis} tersedia tetapi sebagian besar kosong, mencoba jenis berikutnya.")
    riwayat_jenis.append({"jenis": jenis, "hasil": "sebagian besar kosong"})

if data_mentah is None:
    raise RuntimeError("Data belum berhasil diunduh. Coba lagi nanti atau periksa sambungan internet.")

# Simpan respons mentah apa adanya, lalu catatan pengambilan
nama_mentah = f"geomagnet_{KODE_STASIUN}_{jenis_dipakai}_{WAKTU_MULAI:%Y%m%d}_{WAKTU_SELESAI:%Y%m%d}.json"
jalur_mentah = MENTAH / nama_mentah
jalur_mentah.write_bytes(jawaban_dipakai.content)

metadata_pengambilan = {
    "waktu_unduh_utc": pd.Timestamp.now(tz="UTC").isoformat(),
    "url_akhir": jawaban_dipakai.url,
    "parameter": parameter_dipakai,
    "status_http": jawaban_dipakai.status_code,
    "berkas_mentah": str(jalur_mentah),
    "ukuran_byte": jalur_mentah.stat().st_size,
    "sha256": sha256_berkas(jalur_mentah),
    "jenis_data_dipakai": jenis_dipakai,
    "riwayat_jenis_data": riwayat_jenis,
}
jalur_metadata = HASIL / "metadata_pengambilan.json"
jalur_metadata.write_text(json.dumps(metadata_pengambilan, indent=2, ensure_ascii=False), encoding="utf-8")

# Tabel hasil normalisasi (belum dibersihkan)
data_mentah.to_csv(HASIL / "tabel_mentah_normal.csv", index_label="waktu_utc")

info_stasiun = isi_dipakai.get("metadata", {}).get("intermagnet", {}).get("imo", {})
print(f"Jenis data yang dipakai: {jenis_dipakai}")
print(f"Stasiun: {info_stasiun.get('name', NAMA_STASIUN)}")
print(f"Koordinat (bujur, lintang, ketinggian): {info_stasiun.get('coordinates')}")
print(f"Berhasil mengunduh {len(data_mentah):,} baris data.")
print("Respons mentah tersimpan di:", jalur_mentah)
print("Catatan pengambilan tersimpan di:", jalur_metadata)
display(data_mentah.head())
```

## 6. Periksa mutu data (R-QC-001 dan R-QC-002)

Tujuan pemeriksaan bukan memperbaiki semua data, melainkan membuat kondisi data terlihat. Amatan asli tidak diubah. Setiap aturan menghasilkan satu kolom tanda, dan nilai yang dikeluarkan dari analisis ditaruh pada kolom terpisah.

Pemeriksaan yang dilakukan:

- Nilai kosong, termasuk kode 88888 dan 99999 yang dipakai format IAGA untuk nilai hilang.
- Waktu ganda. Kemunculan pertama dipertahankan dan jumlahnya dicatat.
- Batas fisik. Medan magnet bumi di permukaan berkisar sekitar 25.000 sampai 65.000 nT, sehingga batas longgar dipakai untuk F dan H. Nilai di luar batas ditandai dan tidak dipakai dalam analisis.
- Perubahan mendadak. Perubahan H antarmenit yang besar hanya ditandai untuk ditinjau. Perubahan besar belum tentu salah ukur, sebab badai asli juga menghasilkan perubahan cepat, sehingga nilainya tetap dipakai.
- Celah waktu. Interval yang diharapkan ditetapkan dari spesifikasi, yaitu satu menit, bukan ditebak dari data. Setiap celah dilaporkan awal, akhir, lama, dan jumlah menit yang hilang.

Komponen H dihitung setelah penggantian kode nilai hilang.

```python
amatan = data_mentah.copy()

# Transformasi yang dilakukan dicatat jumlahnya
jumlah_kode_hilang = int(amatan.isin([88888, 99999]).sum().sum())
amatan = amatan.replace([88888, 99999], np.nan).sort_index()

jumlah_waktu_ganda = int(amatan.index.duplicated().sum())
amatan = amatan[~amatan.index.duplicated(keep="first")]

# Susun ulang pada grid satu menit agar menit yang hilang terlihat
indeks_penuh = pd.date_range(WAKTU_MULAI, WAKTU_SELESAI, freq="min")
menit_diterima = int(amatan.index.isin(indeks_penuh).sum())
tabel = amatan.reindex(indeks_penuh)
tabel["H"] = np.hypot(tabel["X"], tabel["Y"])

# Tanda mutu, satu kolom untuk satu aturan
flag = pd.DataFrame(index=indeks_penuh)
flag["kosong_H"] = tabel["H"].isna()
flag["luar_batas_F"] = tabel["F"].notna() & ~tabel["F"].between(*BATAS_F)
flag["luar_batas_H"] = tabel["H"].notna() & ~tabel["H"].between(*BATAS_H)
flag["lonjakan_H"] = tabel["H"].diff().abs() > AMBANG_LONJAKAN

# Nilai yang dipakai analisis. Amatan asli (H dan F) tetap utuh.
tabel["H_dipakai"] = tabel["H"].where(~flag["luar_batas_H"])
tabel["F_dipakai"] = tabel["F"].where(~flag["luar_batas_F"])

# Tabel celah waktu untuk komponen H
kosong = tabel["H_dipakai"].isna()
kelompok = (kosong != kosong.shift()).cumsum()
daftar_celah = []
for _, bagian in kosong.groupby(kelompok):
    if bagian.iloc[0]:
        daftar_celah.append({
            "mulai_utc": bagian.index[0],
            "selesai_utc": bagian.index[-1],
            "lama_menit": len(bagian),
            "menit_hilang": len(bagian),
        })
tabel_celah = pd.DataFrame(daftar_celah, columns=["mulai_utc", "selesai_utc", "lama_menit", "menit_hilang"])

ringkasan_mutu = {
    "menit_seharusnya": len(indeks_penuh),
    "menit_diterima": menit_diterima,
    "kelengkapan_H_persen": round(100 * tabel["H_dipakai"].notna().mean(), 2),
    "kode_hilang_diganti": jumlah_kode_hilang,
    "waktu_ganda_dibuang": jumlah_waktu_ganda,
    "menit_kosong_H": int(flag["kosong_H"].sum()),
    "menit_luar_batas_F": int(flag["luar_batas_F"].sum()),
    "menit_luar_batas_H": int(flag["luar_batas_H"].sum()),
    "menit_lonjakan_H_ditandai": int(flag["lonjakan_H"].sum()),
    "jumlah_celah": len(tabel_celah),
}
display(pd.Series(ringkasan_mutu, name="nilai", dtype=object))

if len(tabel_celah) == 0:
    print("Tidak ada celah waktu pada komponen H.")
else:
    display(tabel_celah)

if tabel["H_dipakai"].notna().sum() == 0:
    raise ValueError("Tidak ada data valid setelah pemeriksaan mutu.")

# Simpan tanda, tabel celah, dan laporan mutu
flag.to_csv(HASIL / "flag_mutu.csv", index_label="waktu_utc")
tabel_celah.to_csv(HASIL / "tabel_celah.csv", index=False)
pd.Series(ringkasan_mutu, name="nilai").to_csv(HASIL / "ringkasan_mutu.csv", header=True)

laporan = [
    "# Laporan Mutu Data Geomagnet",
    "",
    f"Stasiun {NAMA_STASIUN} ({KODE_STASIUN}), jenis data {jenis_dipakai}, interval {INTERVAL_DETIK} detik.",
    "",
    "## Temuan faktual",
    "",
]
laporan += [f"- {nama}: {nilai}" for nama, nilai in ringkasan_mutu.items()]
laporan += [
    "",
    "## Catatan interpretasi yang disarankan",
    "",
    "- Tanda lonjakan hanya meminta peninjauan. Nilainya tidak dibuang karena badai asli juga menghasilkan perubahan cepat.",
    "- Amatan asli tidak diubah. Nilai di luar batas fisik dikeluarkan hanya pada kolom H_dipakai dan F_dipakai.",
]
(HASIL / "laporan_mutu.md").write_text("\n".join(laporan), encoding="utf-8")
print("Laporan mutu tersimpan di:", HASIL / "laporan_mutu.md")
```

## 9. Anomali komponen horizontal, awal badai, dan puncak badai (R-CAL-002)

Anomali dihitung sebagai selisih H terhadap rata-rata H pada jendela garis dasar. Nilai negatif berarti medan horizontal lebih lemah daripada kondisi tenang. Periode garis dasar, aturan data kosong (menit kosong dilewati), dan satuan (nT) dinyatakan jelas, dan tabel garis dasar disimpan bersama deret anomalinya. Jumlah menit dan simpangan baku garis dasar ikut dicatat supaya pembaca tahu seberapa kecil perubahan yang masih berarti.

Awal badai diperkirakan dari kenaikan H terbesar dalam lima menit pada hari 10 Mei. Penentuan ini bersifat pendekatan dan bukan penetapan resmi waktu sudden commencement.

Anomali di sini masih memuat sisa variasi harian, sehingga tidak persis sama dengan indeks Dst yang sudah menghilangkan variasi tersebut.

```python
dasar = tabel.loc[DASAR_MULAI:DASAR_SELESAI, "H_dipakai"]
garis_dasar_H = dasar.mean()
tabel["anomali_H"] = tabel["H_dipakai"] - garis_dasar_H

tabel_dasar = pd.DataFrame({
    "periode_mulai_utc": [DASAR_MULAI],
    "periode_selesai_utc": [DASAR_SELESAI],
    "menit_valid": [int(dasar.notna().sum())],
    "rata_rata_H_nT": [round(garis_dasar_H, 2)],
    "simpangan_baku_nT": [round(dasar.std(), 2)],
})
display(tabel_dasar)
tabel_dasar.to_csv(HASIL / "tabel_garis_dasar.csv", index=False)
tabel[["H_dipakai", "anomali_H"]].to_csv(HASIL / "anomali_H.csv", index_label="waktu_utc")

# Puncak badai: anomali paling negatif
waktu_puncak = tabel["anomali_H"].idxmin()
nilai_puncak = tabel["anomali_H"].min()

# Awal badai: kenaikan H terbesar dalam lima menit pada hari pencarian
kenaikan_lima_menit = tabel["H_dipakai"].diff(5).loc[CARI_MULAI:CARI_SELESAI]
waktu_awal = kenaikan_lima_menit.idxmax()
nilai_awal = kenaikan_lima_menit.max()

print(f"Perkiraan awal badai: {waktu_awal:%d-%m-%Y %H:%M} UTC, kenaikan {nilai_awal:.1f} nT dalam 5 menit")
print(f"Puncak badai: {waktu_puncak:%d-%m-%Y %H:%M} UTC, anomali {nilai_puncak:.1f} nT")
print(f"Puncak badai dalam WITA: {waktu_puncak.tz_convert('Asia/Makassar'):%d-%m-%Y %H:%M}")

fig, ax = plt.subplots(figsize=(11, 4.8))
ax.plot(tabel.index, tabel["anomali_H"], linewidth=0.9)
ax.axhline(0, color="black", linewidth=0.8)
ax.axvspan(DASAR_MULAI, DASAR_SELESAI, color="gray", alpha=0.2, label="Jendela garis dasar")
ax.axvline(waktu_awal, color="tab:green", linestyle="--", label="Perkiraan awal badai")
ax.scatter([waktu_puncak], [nilai_puncak], color="tab:red", zorder=3, label="Puncak badai")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%m\n%H:%M"))
ax.set_xlabel("Waktu (UTC)")
ax.set_ylabel("Anomali komponen H (nT)")
ax.set_title(f"Anomali Komponen Horizontal di {NAMA_STASIUN} terhadap Garis Dasar Malam Hari")
ax.legend()
plt.tight_layout()
simpan_gambar(fig, "02_anomali_H.png")
plt.show()
```

## 10. Laju perubahan medan

Laju perubahan medan menunjukkan seberapa cepat H berubah dari menit ke menit. Perubahan yang cepat penting karena dapat menimbulkan arus induksi pada jaringan listrik dan pipa panjang. Nilai di sini dihitung dari satu stasiun dan tidak dapat langsung dipakai untuk menilai risiko pada suatu jaringan tertentu.

```python
laju_mutlak = tabel["H_dipakai"].diff().abs()     # nT per menit

waktu_laju = laju_mutlak.idxmax()
nilai_laju = laju_mutlak.max()
print(f"Laju perubahan terbesar: {nilai_laju:.1f} nT per menit pada {waktu_laju:%d-%m-%Y %H:%M} UTC")

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(laju_mutlak.index, laju_mutlak, linewidth=0.7)
ax.axhline(AMBANG_LONJAKAN, color="tab:orange", linestyle=":", label="Ambang tinjauan")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%m\n%H:%M"))
ax.set_xlabel("Waktu (UTC)")
ax.set_ylabel("Laju perubahan H (nT per menit)")
ax.set_title(f"Laju Perubahan Komponen Horizontal di {NAMA_STASIUN}")
ax.legend()
plt.tight_layout()
simpan_gambar(fig, "03_laju_perubahan_H.png")
plt.show()
```

## 11. Ringkasan per hari

Tabel memuat anomali terendah, anomali tertinggi, dan rentangnya pada setiap hari UTC. Jumlah menit valid dan kelengkapan dicantumkan agar terlihat berapa banyak data di balik setiap baris. Satu hari UTC tidak sama dengan satu hari setempat, sehingga badai yang melewati pergantian hari terbagi pada dua baris.

```python
grup = tabel["anomali_H"].groupby(tabel.index.date)
harian = grup.agg(["min", "max", "count"]).round(1)
harian["rentang"] = (harian["max"] - harian["min"]).round(1)
harian["kelengkapan"] = (100 * harian["count"] / 1440).round(1)
harian.columns = ["Minimum (nT)", "Maksimum (nT)", "Menit valid", "Rentang (nT)", "Kelengkapan (%)"]
harian.index.name = "Tanggal (UTC)"
display(harian)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar([str(t) for t in harian.index], harian["Rentang (nT)"])
ax.set_xlabel("Tanggal (UTC)")
ax.set_ylabel("Rentang anomali H (nT)")
ax.set_title(f"Rentang Anomali Harian Komponen H di {NAMA_STASIUN}")
ax.grid(axis="y", alpha=0.3)
plt.xticks(rotation=30)
plt.tight_layout()
simpan_gambar(fig, "04_rentang_harian.png")
plt.show()

harian.to_csv(HASIL / "ringkasan_harian.csv")
```

## 12. Optional check, verifikasi akhir

Pemeriksaan pertama memastikan tabel tidak kosong dan nilainya masuk akal. Pemeriksaan kedua membandingkan anomali H di Guam dengan indeks Dst yang dihitung USGS. Kedua deret dirata-ratakan per jam lebih dulu.

Perbandingan ini adalah verifikasi terhadap pembanding yang dinyatakan, bukan validasi fisika. Kesesuaian yang baik menandakan pengolahan sudah benar, tetapi nilainya tidak akan identik karena Dst menggabungkan beberapa stasiun dan sudah menghilangkan variasi harian. Bila data Dst tidak dapat diambil, bagian ini hanya memberi pesan dan tidak menghentikan notebook.

```python
# Pemeriksaan konsistensi dasar
assert not tabel.empty, "Tabel kosong"
assert tabel.index.is_monotonic_increasing
assert not tabel.index.duplicated().any()
assert tabel["H_dipakai"].dropna().between(*BATAS_H).all(), "Nilai H di luar batas"
assert tabel["anomali_H"].notna().any()
print("Pemeriksaan konsistensi dasar: lulus")

# Pembanding dengan indeks Dst dari USGS
hasil_dst = {"status": "tidak tersedia"}
try:
    parameter_dst = {
        "id": "USGS",
        "starttime": teks_waktu(WAKTU_MULAI),
        "endtime": teks_waktu(WAKTU_SELESAI),
        "elements": "DST",
        "sampling_period": INTERVAL_DETIK,
        "format": "json",
    }
    tabel_dst = ubah_ke_tabel(ambil(parameter_dst).json())
    dst = tabel_dst.iloc[:, 0]

    pasangan = pd.concat(
        [tabel["anomali_H"].resample("1h").mean(), dst.resample("1h").mean()],
        axis=1,
        keys=["anomali_H", "Dst"],
    ).dropna()

    korelasi = pasangan["anomali_H"].corr(pasangan["Dst"])
    waktu_dst_min = pasangan["Dst"].idxmin()
    hasil_dst = {
        "status": "tersedia",
        "korelasi": round(float(korelasi), 3),
        "Dst_terendah_nT": round(float(pasangan["Dst"].min()), 1),
        "waktu_Dst_terendah_utc": waktu_dst_min.isoformat(),
        "anomali_H_terendah_per_jam_nT": round(float(pasangan["anomali_H"].min()), 1),
        "jumlah_jam_dibandingkan": int(len(pasangan)),
    }
    print(f"Korelasi anomali H dan Dst (rata-rata per jam): {korelasi:.2f}")
    print(f"Dst terendah: {pasangan['Dst'].min():.0f} nT pada {waktu_dst_min:%d-%m-%Y %H:%M} UTC")
    print(f"Anomali H terendah (per jam): {pasangan['anomali_H'].min():.0f} nT")

    pasangan.to_csv(HASIL / "perbandingan_dst.csv", index_label="waktu_utc")

    fig, ax = plt.subplots(figsize=(11, 4.5))
    ax.plot(pasangan.index, pasangan["anomali_H"], label=f"Anomali H di {NAMA_STASIUN}")
    ax.plot(pasangan.index, pasangan["Dst"], label="Indeks Dst (USGS)")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%m\n%H:%M"))
    ax.set_xlabel("Waktu (UTC)")
    ax.set_ylabel("nT")
    ax.set_title("Perbandingan Anomali H dengan Indeks Dst")
    ax.legend()
    plt.tight_layout()
    simpan_gambar(fig, "05_perbandingan_dst.png")
    plt.show()
except Exception as galat:
    hasil_dst = {"status": f"tidak tersedia: {galat}"}
    print("Pembanding Dst belum dapat diambil:", galat)
```

## 15. Manifes hasil (R-PUB-001)

Manifes mencatat nomor jalan, waktu, masukan, konfigurasi, recipe dan prompt, lingkungan perangkat lunak, nomor commit Git bila ada, serta semua berkas keluaran lengkap dengan ukuran dan sidik jarinya. Dengan manifes ini setiap angka dan grafik dapat ditelusuri ke proses yang menghasilkannya. Status eksekusi ditulis di sini oleh proses yang benar-benar berjalan.

```python
waktu_jalan = pd.Timestamp.now(tz="UTC")
id_jalan = f"GMG-{waktu_jalan:%Y%m%d-%H%M%S}"


def nomor_commit_git():
    try:
        hasil = subprocess.run(["git", "rev-parse", "HEAD"], capture_output=True, text=True, timeout=5)
        return hasil.stdout.strip() if hasil.returncode == 0 else "tidak tersedia"
    except Exception:
        return "tidak tersedia"


berkas_keluaran = []
for jalur in sorted(OUT_DIR.rglob("*")):
    if jalur.is_file() and jalur.name != "manifes_hasil.json":
        berkas_keluaran.append({
            "jalur": str(jalur),
            "ukuran_byte": jalur.stat().st_size,
            "sha256": sha256_berkas(jalur),
        })

manifes = {
    "id_jalan": id_jalan,
    "waktu_jalan_utc": waktu_jalan.isoformat(),
    "status_eksekusi": "dijalankan secara lokal; data diambil langsung dari layanan USGS",
    "id_recipe": ["R-GMG-001", "R-RET-001", "R-QC-001", "R-QC-002", "R-CAL-002", "R-VIS-001", "R-PUB-001"],
    "id_prompt": "P-GMG-001",
    "perintah": "dijalankan dari notebook Aschani_H061241019_PROJECT_3.ipynb",
    "konfigurasi": {
        "stasiun": KODE_STASIUN,
        "mulai_utc": WAKTU_MULAI.isoformat(),
        "selesai_utc": WAKTU_SELESAI.isoformat(),
        "unsur": UNSUR,
        "interval_detik": INTERVAL_DETIK,
        "garis_dasar_utc": [DASAR_MULAI.isoformat(), DASAR_SELESAI.isoformat()],
        "batas_F_nT": list(BATAS_F),
        "batas_H_nT": list(BATAS_H),
        "ambang_lonjakan_nT_per_menit": AMBANG_LONJAKAN,
    },
    "masukan": metadata_pengambilan,
    "ringkasan_mutu": ringkasan_mutu,
    "hasil_utama": {
        "garis_dasar_H_nT": round(float(garis_dasar_H), 2),
        "awal_badai_perkiraan_utc": waktu_awal.isoformat(),
        "kenaikan_awal_5menit_nT": round(float(nilai_awal), 2),
        "puncak_badai_utc": waktu_puncak.isoformat(),
        "anomali_puncak_nT": round(float(nilai_puncak), 2),
        "laju_terbesar_nT_per_menit": round(float(nilai_laju), 2),
        "pembanding_dst": hasil_dst,
    },
    "lingkungan": {
        "python": sys.version.split()[0],
        "sistem": platform.platform(),
        "numpy": np.__version__,
        "pandas": pd.__version__,
        "matplotlib": matplotlib.__version__,
        "requests": requests.__version__,
    },
    "git_commit": nomor_commit_git(),
    "berkas_keluaran": berkas_keluaran,
}

jalur_manifes = HASIL / "manifes_hasil.json"
jalur_manifes.write_text(json.dumps(manifes, indent=2, ensure_ascii=False, default=str), encoding="utf-8")

print("Nomor jalan:", id_jalan)
print("Jumlah berkas keluaran tercatat:", len(berkas_keluaran))
print("Manifes tersimpan di:", jalur_manifes)
```
